```python
import pandas as pd
```

Starting our compensation inclusive regressions for robustness checking.

```python
comp1 = pd.read_csv("csv_data/COMP1.csv")
comp1.columns
```

```python
gdp_growth = {
    2015: 2.9, 2016: 1.8, 2017: 2.5, 2018: 3.0,
    2019: 2.6, 2020: -2.2, 2021: 6.1, 2022: 2.5,
    2023: 2.9, 2024: 2.8, 2025: 2.1
}

comp1['gdp_growth'] = comp1['fyear'].map(gdp_growth)
```

```python
print(comp1.columns.tolist())
print(comp1.shape)
```

```python
comp1 = comp1.dropna()
print(comp1.shape)
```

```python
# remove overage CEOs (> 100 years old, not included in sample)
comp1 = comp1[comp1['ceo_age'] < 100]
print(comp1.shape)
```

```python
from linearmodels.panel import PanelOLS

panel_comp = comp1.set_index(['gvkey', 'fyear'])

model_comp_baseline = PanelOLS.from_formula(
    '''adjusted_roa ~ mba_factor + phd_factor + md_factor + 
       masters_factor + ug_factor +
       pct_equity + log_tdc1 +
       firm_size + equity_capital + charter_value + 
       retained_earnings + rd_intensity + volatility +
       ceo_age + CEO_gender + gdp_growth +
       EntityEffects + TimeEffects''',
    data=panel_comp,
    drop_absorbed=True,
    check_rank=False
)

result_comp_baseline = model_comp_baseline.fit(cov_type='clustered', cluster_entity=True)
print(result_comp_baseline.summary)
```
